```python
# Step 1: Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Step 2: Upload CSV file
uploaded = files.upload()

# Step 3: Read entire dataset with correct encoding
# Try UTF-8 first, fallback to ISO-8859-1 if error occurs
try:
    df = pd.read_csv("task1_dataset.csv", encoding="utf-8")
except UnicodeDecodeError:
    df = pd.read_csv("task1_dataset.csv", encoding="ISO-8859-1")

# Display the full dataset
print("----- Entire Dataset -----")
print(df.to_string())   # shows all rows and columns

# Step 4: Histogram for continuous variable (Age)
plt.figure(figsize=(8,6))
plt.hist(df["Age"], bins=10, color="skyblue", edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# Step 5: Bar chart for categorical variable (Gender)
plt.figure(figsize=(6,5))
df["Gender"].value_counts().plot(kind="bar", color=["pink","lightblue"])
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()

# Step 6: Bar chart for another categorical variable (City)
plt.figure(figsize=(8,6))
df["City"].value_counts().plot(kind="bar", color="orange")
plt.title("City Distribution")
plt.xlabel("City")
plt.ylabel("Count")
plt.show()
```
